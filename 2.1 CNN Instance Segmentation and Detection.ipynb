```python
import os
from ultralytics import YOLO
```

```python
# There are other segmentation models that can be imported or we can build it from scratch on PyTorch
model = YOLO('yolov8n-seg.pt')
```

```python
# Verify the dataset
model.data = '/content/gdrive/MyDrive/GitHub Deep Learning for Materials/dataunzip/data/config.yaml'
```

```python
# Train the model
results = model.train(data=os.path.join(ROOT_DIR, "config.yaml"), epochs=50, project = '/content/gdrive/MyDrive/GitHub Deep Learning for Materials', patience = 5, imgsz = 640)  # train the model
```

```python
# Directory containing images for inference
image_dir_val = '/content/gdrive/MyDrive/imagesTest/'  # Change this to your test images folder
```

```python
# Run predictions, generating some clean images without bounding boxes and labels 
results = model.predict(source=image_dir_val, save=True, save_txt=True, save_conf=True, imgsz=640, show_conf = False, show_boxes = False, retina_masks = True, show_labels = False)
```

```python
# Run predictions, generating some images with bounding boxes and labels. 
results = model.predict(
    source=image_dir_val,
    save=True,
    save_txt=True,
    save_conf=True,
    show_conf = False,
    imgsz=640,
    visualize=False
)
```

```python
# Visualize the clean predictions

import os
import matplotlib.pyplot as plt
import matplotlib.patches as patches

# Get the output directory where YOLO saved the predicted images
output_dir = "/content/gdrive/MyDrive/GitHub Deep Learning for Materials/train54"  # Automatically set by YOLO after prediction

# List all predicted image files
image_paths = [os.path.join(output_dir, f) for f in os.listdir(output_dir) if f.lower().endswith(('.png', '.jpg', '.jpeg'))]
image_paths.sort()  # Optional: ensure consistent display order

# Display 3 images per row with black borders
for i in range(0, len(image_paths), 3):
    fig, axes = plt.subplots(1, 3, figsize=(18, 6))
    for j in range(3):
        ax = axes[j]
        if i + j < len(image_paths):
            img = plt.imread(image_paths[i + j])
            ax.imshow(img)
            ax.axis('off')

            # Add black border using Rectangle patch
            rect = patches.Rectangle(
                (0, 0), 1, 1, transform=ax.transAxes,
                linewidth=2, edgecolor='black', facecolor='none'
            )
            ax.add_patch(rect)
        else:
            ax.axis('off')  # Hide unused subplot
    plt.tight_layout()
    plt.show()
```
